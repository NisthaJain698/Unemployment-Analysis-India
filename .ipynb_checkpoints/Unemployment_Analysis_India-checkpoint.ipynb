## Visualizations

```python
sns.histplot(df1['Estimated Unemployment Rate (%)'], bins=30, kde=True)
plt.title("Distribution of Unemployment Rate")
plt.xlabel("Unemployment Rate (%)")
plt.ylabel("Frequency")
plt.show()
```

### Insights
- Most unemployment rates are concentrated between 0% and 15%.
- The distribution is positively skewed, indicating a few extreme unemployment spikes.
- Very high unemployment values were observed during the Covid-19 lockdown period.
- The presence of outliers suggests that certain states experienced severe economic disruption.

```python
sns.barplot(
    x='Area',
    y='Estimated Unemployment Rate (%)',
    data=df1
)

plt.title("Average Unemployment Rate: Rural vs Urban")
plt.show()
```

### Insights
- Urban unemployment was higher than rural unemployment.
- Urban sectors such as retail, hospitality, and services were more affected during Covid-19.
- Rural employment showed relatively greater stability compared to urban areas.
- The results indicate stronger economic vulnerability in urban labor markets.

```python
monthly_trend = df1.groupby('Date')['Estimated Unemployment Rate (%)'].mean()

monthly_trend.plot()
plt.title("Unemployment Rate Over Time")
plt.ylabel("Average Unemployment Rate (%)")
plt.show()
```

### Insights
- Unemployment remained relatively stable throughout 2019 and early 2020.
- A sharp increase was observed during April and May 2020.
- The sudden spike indicates severe economic disruption during the Covid-19 lockdown period.
- Unemployment levels started declining after lockdown restrictions were gradually eased.

# Covid Time Analysis

```python
plt.figure(figsize=(14,6))

plt.plot(monthly_trend.index,
         monthly_trend.values,
         marker='o',
         linewidth=2)

plt.axvline(pd.to_datetime('2020-03-01'),
            color='red',
            linestyle='--',
            label='Covid Lockdown Begins')

plt.title("Unemployment Trend Before and During Covid-19")
plt.xlabel("Date")
plt.ylabel("Average Unemployment Rate (%)")

plt.legend()

plt.xticks(rotation=45)

plt.show()
```

### Insights
- The unemployment trend changed dramatically after the Covid-19 lockdown began in March 2020.
- April 2020 recorded the highest unemployment spike in the dataset.
- The graph clearly shows the direct impact of lockdown measures on employment levels.
- Economic activities and labor markets were heavily disrupted during the pandemic.

```python
before_covid = df1[df1['Date'] < '2020-03-01']
during_covid = df1[df1['Date'] >= '2020-03-01']

print("Average Unemployment Before Covid:",
      before_covid['Estimated Unemployment Rate (%)'].mean())

print("Average Unemployment During Covid:",
      during_covid['Estimated Unemployment Rate (%)'].mean())
```

covid_comparison = pd.DataFrame({
    'Period': ['Before Covid', 'During Covid'],
    'Unemployment Rate': [
        before_covid['Estimated Unemployment Rate (%)'].mean(),
        during_covid['Estimated Unemployment Rate (%)'].mean()
    ]
})

sns.barplot(
    x='Period',
    y='Unemployment Rate',
    data=covid_comparison
)

plt.title("Average Unemployment Rate: Before vs During Covid")
plt.show()

# Rural vs Urban Analysis

```python
area_unemployment = df1.groupby('Area')['Estimated Unemployment Rate (%)'].mean()

print(area_unemployment)
```

```python
sns.barplot(
    x='Area',
    y='Estimated Unemployment Rate (%)',
    data=df1
)

plt.title("Average Unemployment Rate: Rural vs Urban")
plt.xlabel("Area")
plt.ylabel("Unemployment Rate (%)")

plt.show()
```

### Insights
- Average unemployment increased significantly during the Covid-19 period.
- The unemployment rate nearly doubled compared to the pre-pandemic period.
- The results highlight the severe economic consequences of nationwide lockdown restrictions.
- Covid-19 created one of the largest employment disruptions in recent years.

```python
area_trend = df1.groupby(['Date', 'Area'])['Estimated Unemployment Rate (%)'].mean().reset_index()
plt.figure(figsize=(14,6))

sns.lineplot(
    data=area_trend,
    x='Date',
    y='Estimated Unemployment Rate (%)',
    hue='Area',
    marker='o'
)

plt.title("Rural vs Urban Unemployment Trend")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=0)

plt.show()
```

### Insights
- Urban unemployment remained consistently higher than rural unemployment.
- Both rural and urban areas experienced sharp unemployment spikes during Covid-19.
- Urban unemployment increased more rapidly during lockdown periods.
- Rural unemployment showed relatively better recovery stability after the Covid spike.

```python
before_area = before_covid.groupby('Area')['Estimated Unemployment Rate (%)'].mean()
during_area = during_covid.groupby('Area')['Estimated Unemployment Rate (%)'].mean()

covid_area_comparison = pd.DataFrame({
    'Before Covid': before_area,
    'During Covid': during_area
})

print(covid_area_comparison)
```

```python
covid_area_comparison.plot(kind='bar')

plt.title("Rural vs Urban Unemployment: Before and During Covid")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=0)

plt.show()
```

### Insights
- Unemployment increased significantly in both rural and urban areas during Covid-19.
- Urban regions experienced higher unemployment compared to rural regions.
- The pandemic had a stronger impact on service-based urban economies.
- Rural areas were affected as well, but showed comparatively lower unemployment levels.

# State Wise Analysis


```python
state_avg = df1.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

print(state_avg)
```

```python
top_states = state_avg.head(10)

plt.figure(figsize=(12,6))

sns.barplot(
    y=top_states.values,
    x=top_states.index
)

plt.title("Top 10 States with Highest Average Unemployment")
plt.ylabel("Average Unemployment Rate (%)")
plt.xlabel("State")

plt.show()
```

### Insights
- Tripura recorded the highest average unemployment rate among all states.
- Haryana and Jharkhand also showed consistently high unemployment levels.
- Northern and eastern regions displayed relatively higher unemployment rates.
- These states may require stronger employment and economic support policies.

```python
lowest_states = state_avg.tail(10)

plt.figure(figsize=(12,6))

sns.barplot(
    
    x=lowest_states.index,
    y=lowest_states.values
)

plt.title("States with Lowest Average Unemployment")
plt.xlabel("State")
plt.xlabel("Average Unemployment Rate (%)")


plt.xticks(rotation=40)


plt.show()
```

### Insights
- Meghalaya recorded the lowest average unemployment rate.
- Odisha, Assam, and Uttarakhand also maintained relatively lower unemployment levels.
- These states showed comparatively stable labor market conditions.
- Lower unemployment may indicate stronger local employment absorption or smaller urban dependency.

```python
top5_states = state_avg.head(5).index

top5_data = df1[df1['Region'].isin(top5_states)]

plt.figure(figsize=(14,6))

sns.lineplot(
    data=top5_data,
    x='Date',
    y='Estimated Unemployment Rate (%)',
    hue='Region',
    marker='o'
)

plt.title("Unemployment Trends for Top 5 Most Affected States")
plt.xlabel("Date")
plt.ylabel("Unemployment Rate (%)")

plt.xticks(rotation=45)

plt.show()
```

### Insights
- Tripura and Haryana consistently recorded very high unemployment rates.
- All top affected states showed major unemployment spikes during the Covid-19 period.
- April and May 2020 marked the peak unemployment months for most states.
- The trends indicate that Covid-19 intensified already existing employment challenges in several regions.

```python
pivot_table = df1.pivot_table(
    values='Estimated Unemployment Rate (%)',
    index='Region',
    columns='Month_Name'
)
```

# Seasonal Trend


```python
monthly_avg = df1.groupby('Date')['Estimated Unemployment Rate (%)'].mean()
print(monthly_avg)
```

```python
month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

monthly_avg = df1.groupby('Month_Name')['Estimated Unemployment Rate (%)'].mean()

monthly_avg = monthly_avg.reindex(month_order)

plt.figure(figsize=(14,6))

sns.barplot(
    x=monthly_avg.index,
    y=monthly_avg.values
)

plt.title("Average Unemployment Rate by Month")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)

plt.show()
```

### Insights
- April recorded the highest average unemployment rate among all months.
- May also showed elevated unemployment levels due to continued lockdown effects.
- Most months before Covid-19 maintained relatively stable unemployment patterns.
- The monthly analysis clearly reflects the seasonal and pandemic-driven employment disruptions.

```python
plt.figure(figsize=(14,6))

sns.lineplot(
    x=monthly_avg.index,
    y=monthly_avg.values,
    marker='o',
    linewidth=3
)

plt.title("Average Unemployment Rate by Month")
plt.xlabel("Month")
plt.ylabel("Average Unemployment Rate (%)")

plt.xticks(rotation=45)

plt.show()
```

### Insights
- April recorded the highest average unemployment rate across the dataset.
- May also showed elevated unemployment levels due to prolonged lockdown effects.
- Most months before Covid-19 maintained relatively stable unemployment patterns.
- The graph clearly highlights the severe employment disruption caused during the pandemic period.

# Correlation Analysis

```python
correlation = df1[['Estimated Unemployment Rate (%)',
                   'Estimated Employed',
                   'Estimated Labour Participation Rate (%)']].corr()

print(correlation)
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm'
)

plt.title("Correlation Between Employment Variables")

plt.show()
```

### Insights
- Estimated employment shows a negative correlation with unemployment rate.
- Higher employment levels are generally associated with lower unemployment.
- Labour participation rate showed a weak correlation with unemployment rate.
- The variables indicate interconnected labor market behavior during economic disruptions.

## Employment vs Unemployment



```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df1,
    x='Estimated Employed',
    y='Estimated Unemployment Rate (%)',
    hue='Area'
)

plt.title("Employment vs Unemployment Rate")

plt.show()
```

### Insights
- States with lower employment levels generally showed higher unemployment rates.
- The scatter distribution suggests an inverse relationship between employment and unemployment.
- Urban and rural observations are spread differently, indicating varying labor market conditions.
- Several extreme unemployment observations were recorded during the Covid-19 period.
